# Break , Continue, Pass

```python
for i in range(1,11):
    if i == 6:
        break
    print(i)
```

```python
for i in range(1,11):
    if i % 2 == 0:
        continue
    print(i)
```

```python
for i in range(5):
    if i == 2:
        pass # Do nothing
    print(i)
```

# 11) Count, Find, Index 

- count : Count Occurance of Substring
- find  : Find first occurance(-1)
- index : find first occurance (raise error)

```python
text = "Python Programming"

print(text.count("p"))
print(text.count('o'))
print(text.count('gram'))
```

```python
print(text.find('p'))
print(text.find("gram"))
print(text.find("Java"))
```

```python
print(text.index("P"))
print(text.index("gram"))
print(text.index("Java"))
```
