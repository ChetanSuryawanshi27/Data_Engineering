# 1 . Check if a Number is Even or Odd

```python
num = int(input("Enter a Number: "))  # input() takes input from the user , int() converets string to integer

if num % 2 == 0:    # % is the modulus operator  , If num % 2 == 0, the number is divisible by 2.
    print("Even Number")
else:
    print("Odd Number")
```

# 4 . Find the Sum of Numbers from 1 to N

```python
n = int(input("Enter N: "))                # i = 1  total = 0 + 1 = 1
total = 0                                  # i = 2  total = 1 + 2 = 3
                                           # i = 3  total = 3 + 3 = 6
for i in range(1, n + 1):                  # i = 4  total = 6 + 4 = 10
    total = total + i                      # i = 5  total = 10 + 5 = 15
print("Sum: ",total)
```

# 7 . Reverse a Number

```python
num = int(input("Enter a Number: "))

reverse = 0 

while num > 0:                        #Keep repeating the code as long as num is greater than 0
    digit = num % 10                  #The % operator gives the remainder
    reverse = reverse * 10 + digit
    num = num // 10                   # // is floor division  , // 10 = remove the last digit
print("Reverce:", reverse)            # % 10 = take the last digit
                                      # * 10 + digit = build the reversed number
```

# 9 . Check Palindrome Number

```python
num = int(input("Enter a Number: "))  # palindrome number is a number that reads the same forward and backward
                                    # Reverse the digits of the number
original = num                      # Compare the reversed number with the original
reverse = 0                         # If they are the same, it is a palindrome; otherwise, it is not

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit       # * 10 + digit = build the reversed number
    num = num // 10                      # // is floor division

if original == reverse:              
    print("Palindrome")
else:
    print("Not Palindrome")
```

# 14 . Count Even and Odd Numbers in a List

```python
numbers = [10, 15, 20, 25, 30, 35]

even = 0 
odd = 0

for num in numbers:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even Number: ", even)
print("Odd NUmber: ", odd)
```

# 15 . Remove Duplicates from a List

```python
numbers = [10, 20, 10, 30, 20, 40, 30]

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)
print(unique)
```

# 16 . Count Frequency of Each Element

```python
numbers = [1, 2, 2, 3, 3, 3, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```

# 17 . Count Vowels in a String

```python
text = input("Enter a String: ")

vowels = "aeiou"
count = 0

for char in text.lower():   # lower() converts uppercase letters to lowercase
    if char in vowels:
        count += 1 

print("Number of vowels:", count)
```

# 20 . Simple Student Grade Program

```python
marks = []

for i in range(5):
    mark = float(input("Enter mark: "))
    marks.append(mark)

total = sum(marks)
average = total / len(marks)

print("Total:", total)
print("Average:", average)

if average >= 90:
    print("Grade: A")
elif average >= 75:
    print("Grade: B")
elif average >= 60:
    print("Grade: C")
elif average >= 40:
    print("Grade: D")
else:
    print("Grade: F")
    break
```

# 21 . Number Guessing

```python
import random

secret_number = random.randint(1,100)
attempts = 0

print("Enter a Number Between 1 and 100")

while True:
    guess = int(input("Enter Your Guess"))
    attempts += 1

    if guess > secret_number:
        print("Too High!")
    elif guess < secret_number:
        print("Too Low!")
    else:
        print(f"Correct you Guess The Number in : {attempts}")
        break
```
