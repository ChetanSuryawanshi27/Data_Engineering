# 11) List Comprehension

- concised way to create lists using a single line of code

-  syntax : [expression for item in iterable if condition]

```python
squares = []

for x in range(1,6):
    squares.append(x**2)

print(squares)
```

```python
squares = [x**2 for x in range(1,6)]

print( squares)
```

```python
cube = []

for x in range(1,6):
    cube.append(x**3)

print(cube)    
```

```python
cube = [x**3 for x in range(1,6)]

print(cube)
```

```python
even = [ x for x in range(1,11) if x % 2 == 0]

print(even)
```

```python
odd = [y for y in range(1,22) if y % 3 == 0] 

print(odd)
```

# 12) zip function

- combines multiple iterable into tuples

```python
names = ["Pratik","Chetan","Dhruve"]
ages = [24,22,18]

combined = list(zip(names,ages))

print(combined)
```

# unzip

```python
names, ages = zip(*combined)

print(names)
print(ages)
```

```python
subjects = ["Math","Scince","History"]
marks = [70,87,85]
grade = ["B","A+","A"]

merg = list(zip(subjects, marks, grade))
print(merg)
```
