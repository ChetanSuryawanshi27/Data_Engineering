```python
class Parent:
    def __init__(self):
        self.__private_data = "Secure Data"

    def get_private_data(self):  # getter method
        return self.__private_data

class Child(Parent):
    def __init__(self): 
        super().__init__() # call parent Constructor

    def display(self):
        print(f"Data : {self.get_private_data()}")

child = Child()
child.display()
```

# Method Overriding :

- child class provides its  own implementation of a method that exist in parent class 

```python
class Animal:
    def sound(self):
        print("Animale makes sound")

class Dog(Animal):
    def sound(self):
        print("Dog Barks")

class Cat(Animal):
    def sound(self):
        print("Cat Meows")

dog = Dog()
dog.sound()

cat = Cat()
cat.sound()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary 

    def display(self):
        print(f"Name : {self.name}")
        print(f"Salary : {self.salary}")

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name,salary)
        self.department = department

    def display(self):
        super().display()
        print(f"Department : {self.department}")

mgr = Manager("Chetan", 50000, "IT")
mgr.display()
```

# Super Kyeword :

- super() is used to call methods from parent class

- Uses

- call parent constructor 
- call parent method
- access parent attributes

```python
class Parent:
    def show(self):
        print("Parent Show Method")

class Child(Parent):
    def show(self):
        super().show()
        print("Child Show Method")

child = Child()
child.show()
```

```python
class Grandparent:
    def display(self):
        print("Grandfather")

class Parent(Grandparent):
    def display(self):
        super().display()
        print("Parent")

class Child(Parent):
    def display(self):
        super().display()
        print("Child")

child = Child()
child.display()
```

# Super Constructor :

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

class Car(Vehicle):
    def __init__(self , brand, model, fuel_type):
        super().__init__(brand, model) # call parent constructor
        self.fuel_type = fuel_type

car = Car("Maruti", "Swift", "Petrol")
print(car.brand)
print(car.model)
print(car.fuel_type)
```

# Type of  Inheritance :

# 1) Single Inheritance : One child inherits from one parent

```python
class Parent:
    def parent_method(self):
        print("Parent Method")

class Child(Parent):
    def child_method(self):
        print("Child Method")

child = Child()
child.parent_method()
child.child_method()
```

# 2) Multilevel Inheritance : chain of inheritance, child become parent for next class 

```python
class Grandparent:
    def gp_method(self):
        print("Grandparent Method")

class Parent(Grandparent):
    def p_method(srlf):
        print("Parent Method")

class Child(Parent):
    def c_method(self):
        print("Child Method")

child = Child()
child.c_method()
child.p_method()
child.c_method()
```

# 3) Hierarchical Inheritance :

- Multiple child classes inherits from one parent calss

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

class Cat(Animal):
    def meow(self):
        print("Meowing")

class Cow(Animal):
    def moo(self):
        print("Mooing")

dog = Dog()
dog.eat()
dog.bark()

cat = Cat()
cat.eat()
cat.meow()
```

# 5) Multiple Inheritance :

- one child class inherits from multiple parent classes.

```python
class Father:
    def father_method(self):
        print("Father Method")

class Mother:
    def mother_method(self):
        print("Mother method")

class Child(Father, Mother):
    def child_method(self):
        print("Chile Method")

child = Child()
child.father_method()
child.mother_method()
child.child_method()
```

# 5) Hybrid Inheritance :

- combination of two or more types of inheritance

```python
class A:
    def a_method(self):
        print("Method A")

class B(A):
    def b_method(self):
        print("Method B")

class C(A):
    def c_method(self):
        print("Method C")

class D(B,C):
    def d_method(self):
        print("Method D")

d = D()
d.a_method()
d.b_method()
d.c_method()
d.d_method()
```

# Method Overriding :

- same method name in parent class and child class with different implementation

```python
class Shape:
    def area(self):
        print("Calculate Area")

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self): #Overriding
        return self.length * self.width 

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius**2

rect = Rectangle(10,5)
print(rect.area())

circle = Circle(7)
print(circle.area())
```

# Method Overloading :

- python does support traditional method overloading. but we can achive it using default argument

```python
class shape:
    def are(self, radius):
        return 3.14 * radius * radius 

    def area(self, a, b ):
        return a * b 

shape = shape()
shape.area(2,)
shape.area(2.4)
```

```python
class Shape:
    def area(self, a, b=0):
        if b==0:
            return 3.14*a*a
        else:
            return a * b 

shape = Shape()
print(shape.area(2))
print(shape.area(3,4))
```
