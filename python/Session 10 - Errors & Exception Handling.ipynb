# Try - Except - Else Block :

- Else block
- execute only if no Eception occurs in try block.

```python
try:
    num1 = int(input("Enter Number 1:"))
    num2 = int(input("Enter Number 2:"))
    result = num1/num2
except ValueError:
    print("Enter valid Number")
except ZeroDivisionError:
    print("Can not Divide by Zero")
else:
    print(f"Division Successfull : {result}")
```

# Try - Except - Else Finally :

- Finally Block
- Always Executes, whether Exception Occures or not 

```python
def divide_number():
    try:
        num1 = int(input("Enter number 1:"))
        num2 = int(input("Enter number 2:"))
        result = num1/num2
    except ValueError:
        print("Enter valid number")
    except ZeroDivisionError:
        print("You can not divide by zero to any number")
    else:
        print(f"Result : {result}")
        return result
    finally:
        print("Function Execution completed")
```

```python
divide_number()
```

# Creating Custom Exception :

- how to create custom exception :
- create a class that inherits from exception class 

```python
class InvalidAgeError(Exception):
    pass
```

```python
def check_voting_age(age):
    if age <18:
        raise InvalidAgeError("You must be 18 or older to vote")
    else:
        print("you are eligible to vote")
```

```python
check_voting_age(45)
```

```python
try:
    check_voting_age(15)
except InvalidAgeError as e:
    print(f"Error : {e}")
```
