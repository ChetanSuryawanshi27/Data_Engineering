# c) Assignment Operators

 # D) Logical operators

- combine multiple condition

| Operator | Name        | Meaning                                | Example             | Result |
| -------- | ----------- | -------------------------------------- | ------------------- | ------ |
| `and`    | Logical AND | True if both conditions are true       | `(5 > 2 and 8 > 3)` | True   |
| `or`     | Logical OR  | True if at least one condition is true | `(5 > 10 or 8 > 3)` | True   |
| `not`    | Logical NOT | Reverses the result                    | `not(5 > 2)`        | False  |


```python
a = 10 
b = 20
c = 30
```

```python
print(a < b and b < c)
```

```python
print(a > b or b > c)
```

```python
print(not  (a > b) )
```

```python
print(not  (a < b) )
```

# if - elif - else statement:

```python
marks = int(input("Enter your marks :"))

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("grade B")
elif marks >= 70:
    print("Grade C")
elif marks >= 60:
    print("Grade D")
else:
    print("Fail")
```

# Nested if :

```python
age = int(input("Enter your age :"))

if age >=18:
    marks = int(input("Enter your marks :"))
    if marks >= 60:
        print("Eligible for Admission")
    else:
        print("Not Eligible - Low marks")
else:
    print("Not Eligible - Age Below 18")
```

# 4) Loops in python

- Execute a Block of code repeatedly

# A) While Loop :

- Execute as long as True :

# syntax :

# while condition:
#      code to execute
#      update codition

```python
count = 1

while count <= 5:
    print(count)
    count += 1 # count = count + 1
```

```python
num = 5
i = 1

while i <= 10:
    print(num * i)
    i += 1 # i = i + 1
```

```python
count = 1

while True:
    print(count)
    count += 1
    if count > 5:
        break # Breaks the loop
```

```python
num = 0 

while num < 10:
    num += 1
    if num % 2 == 0:  # even numbner Skip
        continue  # Skip Number
    print(num)
```

```python
num = 0 

while num < 10:
    num += 1
    if num % 2 != 0:  # odd numbner Skip
        continue  # Skip Number
    print(num)
```

```python
# while loop with else

count = 1

while count <= 5:
    print(count)
    count += 1
else:
    print("Loop Completed Successfully")
```

# Using range :


```python
for i in range(1,6):  #range(start, end) prints 1 to 5 
    print(i)
```

```python
for i in range(2,11,2):   # range(satrt, end, step)
    print(i)
```

```python
for i in range(3):  # range(end)
    print(i)
```

```python
num = 7

for i in range(1,11):
    print(num * i)
```

```python
for i in range(1,11):
    if i == 6:
        break
    print(i)
```

```python
for i in range(1,11):
    if i % 2 == 0:
        continue
    print(i)
```

```python
for i in range(1,11):
    if i % 2 != 0:
        continue
    print(i)
```

# C) Nested Loop

```python
for i in range(1,4):
    for j in range(1,4):
        print(i , j)
```

```python
# Star pattern

for i in range(1,6):
    for j in range(i):
        print("*", end="")
    print()
```

```python
for i in range(1,10):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# Number Pattern

for i in range(1,6):
    for j in range(1,i + 1):
        print(j, end=" ")
    print()
```
