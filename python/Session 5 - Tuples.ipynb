# 3) Editing Items in a tuple

- Tuples are Immutable
- Tuple cannot be modified after creation

```python
number = (1,2,3,4,5)
number[0] = 100
```

```python
numbers = (1,2,3,4,5)

templist = list(numbers)
templist[4] = 500
numbers = tuple(templist)
print(numbers)
```

```python
numbers = (1,2,3,4,5)
numbers = (100,) + numbers[1:]
print(numbers)
```

```python
numbers = (1,2,3,4,5)
numbers = (100,) + numbers[2:]
print(numbers)
```

```python
my_tuple = (1,2,[3,4,5])
my_tuple[2][0] = 100
print(my_tuple)
```

# 7) Tuple function

- len() - Number of items
- min() - minimum
- max() - maximum
- tuple() - convert to tuple
- count(item) - count occurance
- index(item) - find index first occurance
- sum() - sum of all items
- sorted() - return sorted list

```python
number = (10,50,30,20,40,20)

print(len(number))
print(min(number))
print(max(number))
print(sum(number))
print(number.count(20))
print(number.index(30))
```

```python
print(sorted(number))
```

# 8) Difference between list and Tuple


```python
my_list = [1,2,3]
my_list[0] = 100

print(my_list)
```

```python
my_tuple = (1,2,3)
my_list = list(my_tuple)

my_list[1] = 200
print(my_list)
```

```python
print(my_tuple)
```

```python
import timeit
```

```python
list_time = timeit.timeit("[1,2,3,4,5]" , number = 1000000)
tuple_time = timeit.timeit("(1,2,3,4,5)", number = 1000000)

print("List Creation Time  :" , list_time) # lsit are slower
print("tuple creation Time :" , tuple_time) # tuple are faster
```

```python
import sys

my_list = [1,2,3,4,5]
my_tuple = (1,2,3,4,5)

print("List size" , sys.getsizeof(my_list)) # list takes more memory
print("List size" , sys.getsizeof(my_tuple)) # tuple takes less memory
```

```python
my_list = [1,2,3]
my_tuple = (1,2,3)

print(dir(my_list))
print(dir(my_tuple))
```

# 9) Tuple Unpacking

- Assigning tuple items to multiple variable in one statement

```python
coordinates = (10 , 20)
x , y = coordinates 

print(x , y)
```

```python
name , age , city = ("Chetan", 23 , "Jamner")
print(name,age,city)
```

```python
numbers = (1,2,3,4,5)
first, *middle , last = numbers

print(first)
print(middle)
print(last)
```

```python
a , b , *rest = numbers
print(a)
print(b)
print(rest)
```

# Zip function 

- combine multiple iterables into tuple

```python
names = ("chetan", "priya","pratik")
ages = (23,22,25)

combined = tuple(zip(names,ages))
print(combined)
```

```python
combined = (('chetan', 23), ('priya', 22), ('pratik', 25))
names , ages = zip(*combined)
print()
```
