# Multiple Object

```python
class Car:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def display(self):
        print(f"{self.brand} {self.model} - Rs {self.price}")
```

```python
car1.display()
car2.display()
car3.display()
```

# Method Vs Function :

```python
# Function

def calculate_area(lenght, width):
    return lenght * width

result = calculate_area(10,5)
print(result)
```

```python
# Method 

class Rectangle:
    def __init__(self, lenght, width):
        self.lenght = lenght
        self.width = width

    def calculate_area(self):
            return self.lenght * self.width
```

```python
rect = Rectangle(10,5)
result = rect.calculate_area()
print(result)
```

# Magic Method / Daunder Method :

- Special method with double underscore before and after name.

```python
class Student:
    def __init__(self, name, marks): # Magic Method
        self.name = name
        self.marks = marks

    # String representation
    def __str__(self):  #Magic Method
        return f"Student : {self.name}, Marks: {self.marks}"
```

```python
student1 = Student("Chetan", 75)
print(student1)
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return (self.value + other.value)
```

```python
num1 = Number(10)
num2 = Number(20)

result = num1 + num2
print(result)
```

# Concept of self :

- self represent the current instance of the class. its used to access atteibutes and method.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(id(self))

    def display(self):
        print(f" Name : {self.name}")
        print(f" Age : {self.age}")
```

```python
student1 = Student("Pratik", 23)
student1.display()
print(id(student1))
```

```python
student2 = Student("Chetan",22)
student2.display()
```

```python
class Calculator:

    def __init__(self , num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def multiply(self):
        return self.num1 * self.num2

    def  display_result(self):
        print(f"Sum     : {self.add()}")
        print(f"Product : {self.multiply()}")
```

```python
calc = Calculator(10,5)
calc.display_result()
```
