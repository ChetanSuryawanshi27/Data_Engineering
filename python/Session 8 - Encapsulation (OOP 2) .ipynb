# Mutability of object :

- Mutable objects can be changed after creation. immutable objects can not be cheanged

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
```

```python
student1 = Student("pratik", 75)
print(student1.marks)
print(id(student1))
```

```python
student1.marks = 90
print(student1.marks)
print(id(student1))
```

```python
my_list = [1,2,3] # List are Mutable
print(id(my_list))
my_list.append(4)
print(my_list)
print(id(my_list))
```

```python
x = 10         # integer are mutable
print(id(x))
x = 20
print(id(x))
```

```python
name = "Chetan"        # String are immutable
print(id(name))
name = "pratik"
print(id(name))
```

# 6) Encapsulatiuon:

- Encapsulation means hiding internal details and restricting direct access to date:

# Public attribute

```python
class Student:
    def __init__(self, name, age):
        self.name = name    # Public attribute
        self.age = age      # public attribute
```

```python
student1 = Student("Pratik", 45)
```

```python
print(student1.name)
```

```python
print(student1.age)
```

```python
student1.age = 'hehehe'
```

```python
print(student1.age)
```

```python
print(student1.name)
print(student1.age)
```

# Protected attribute:

```python
class Student:
    def __init__(self, name, age):
        self.name = name   # public attribute
        self._age = age    # protected attribute
```

```python
student1 = Student("Chetan", 23)
```

```python
print(student1.name)
```

```python
print(student1._age)
```

```python
student1._age = 'hehehe'
```

```python
print(student1._age)
```

```python
print(student1.name)
print(student1._age)
```

# Getters & Setters Method :

```python
class Student:
    def __init__(self, name, age):
        self.name = name   # public attribute
        self.__age = age   # private attribute 

    def get_age(self):     # getter method (value ko get karne ke liye)
        return self.__age

    def set_age(self, age):  # setter method (value ko change karne ke liye)
        if type(age) == int:
            if age > 0 :
                self.__age = age
        else:
            print("You Does not change the value")
```

```python
student1 = Student("Chetan", 45)
```

```python
print(student1.name)
```

```python
print(student1.get_age())
```

```python
student1.set_age('heheh')
```

```python
student1.set_age(23)
```

```python
print(student1.get_age())
```

# 7) Collection of objects :

- you can multiply objects in lists,Dictionaries, other collction

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Name:{self.name} , Marks:{self.marks}")
```

```python
s1 = Student("Rahul", 85)
s2 = Student("Ajay", 95)
s3 = Student("Sonali", 75)
```

```python
students = [s1, s2, s3]

for student in students:
    student.display()
```

# 8) Static variable :

- static variables are class variable

```python
class Student:

    school_name = "Bikkad IT"   # Satic variable  (class ke variable)
 
    def __init__(self, name, roll_no):
        self.name = name         # Instance variable  (Object ke variable)
        self.roll_no = roll_no   # Instance variable  (object ke variable)
```

```python
print(Student.school_name)
```

```python
s1 = Student("Ajay", 25)
```

```python
print(s1.name)
print(s1.roll_no)
print(s1.school_name)
```

```python
s1.school_name = "Dhomase IT"   # we can change static variable value  using object name,
                              # but we cannot change it directly
```

```python
print(s1.school_name)
```

```python
print(Student.school_name)   # we change value but its not changed 
                           # we have to use class name for static variable
```

```python
Student.school_name = "Dhomase IT"

print(Student.school_name)
print(s1.school_name)
```

# 9) Static Methd :

```python
class Calculatior:

    @staticmethod   # Decoder
    def add(a,b):
        return a + b

    @staticmethod
    def multiply(a,b):
         return a * b
```

```python
print(Calculatior.add(10,20))
print(Calculatior.multiply(10,20))
```
