# Row Number

```python
w=Window.partitionBy("skills").orderBy(col('salary').desc())
row_number_df=employee_info_df.withColumn('row_number' , row_number().over(w))
```

```python
row_number_df.show(truncate=False)
```

```python
w=Window.partitionBy("city").orderBy(col('salary').desc())
rank_number_df=employee_info_df.withColumn('rank' , rank().over(w))
```

```python
rank_number_df.select("rank","city","salary").show(truncate=False)
```

```python
rank_number_df.show(truncate=False)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import dense_rank, col
```

```python
 w=Window.partitionBy('city').orderBy(col('salary').desc())
 dense_number_df=employee_info_df.withColumn('dense_rank', dense_rank().over(w))
```

# NTILE Divide employee into 2 salary buckets 

```python
w=Window.partitionBy('skills').orderBy(col('salary').desc())
rank_number_ntile_df=employee_info_df.withColumn('bucket', ntile(3).over(w))
```

```python
rank_number_ntile_df.select('bucket','skills','salary').display()
```

```python
w=Window.orderBy(col('salary').desc())
rank_number_ntilewt_df=employee_info_df.withColumn('bucket', ntile(3).over(w))
```

```python
rank_number_ntilewt_df.select('bucket','skills','salary').display()
```

```python
w=Window.partitionBy('skills').orderBy(col('salary').desc())
rank_number_led_df=employee_info_df.withColumn('next_salary', lead('salary').over(w))
rank_number_led_df.select('next_salary','skills','salary').display()
```

```python
w=Window.partitionBy('city').orderBy(col('salary').desc())
rank_number_led_df=employee_info_df.withColumn('next_salary', lead('salary').over(w))
rank_number_led_df.select('next_salary','city','salary').display()
```

```python
w=Window.partitionBy('joining_date').rowsBetween(Window.unboundedPreceding,Window.currentRow)
running_total_led_df=employee_info_df.withColumn('running_total', sum('salary').over(w)) 
```
