# Scope & LEGB Rule :

# What is Scope :

- Scope defines the region in code where a variable is accessible.

# LEGB Rule :

- Local (L) - inside current function
- Enclosing (E) - inside Enclosing function (nested function)
- Global (G) - At module level (outside of function)
- Built-in (B) - Python Built-in Scope

```python
x = 10   # Global variable ( Global Scope ) - (G)

def outer():
    x = 7   # Enclosing Scope (E)

    def inner():
        x = 5   # Local variable ( Local Scope )  (L)
        print("inner", x)

    inner()
    print("outer",x)

outer()
print("Global",x)
```

```python
global_var = "I am Global"  # Global variable ( Global Scope )

def global_function():
    global_var = "I am Global"   # Local variable ( Local Scope )
    print(global_var)

global_function()
print(global_var)
```

```python
# Modify Global Variable :

counter = 0

def increament():
    global counter 
    counter += 1
    print("Counter :", counter)

increament()
increament()
increament()
print(counter)
```

# Enclosing Scope :

- Enclosing Scope exists in nested function.

```python
def outer():
    x = "Enclosing"  # Enclosing Scope

    def inner():
        print(x)

    inner()

outer()
```

```python
x = 100

def outer():
   # x = "Outer" # Enclosing Scope

    def middle():
     #   x = "Middle"  # Enclosing Scope

        def inner():
            print(x)

        inner()

    middle()

outer()
```

# Decorators :

- What is Decorators?
- A decorator is a function that modifies the behaviour of another function without changing its code.

```python
# Basic Decorator :

def my_decorator(func):
    def wrapper():
        print("************")
        func()
        print("************")
    return wrapper

def say_hello():
    print("Hello")

hello = my_decorator(say_hello) 
hello()
```

```python
def my_decorator(func):
    def wrapper():
        print("************")
        func()
        print("************")
    return wrapper
    
@my_decorator
def say_hello():
    print("Hello")

say_hello()
```

# Example - 2 

```python
def decorator(func):
    def wrapper(*args):  # Variable Positional argument
        print("Before function call")
        result = func(*args)
        print("After Function call")
        return result
    return wrapper

@decorator
def add(a,b):
    print(f"Addition {a} + {b} ")
    return a + b

result = add(5,10)
print("Result:",result)
```

```python
# Login Required Decorator:


def login_required(func):
    def wrapper(user):
        if user.get("logged_in"):
            return func(user)
        else:
            print("Please Login First")
    return wrapper

@login_required
def view_profile(user):
    print(f"Welcome {user['name']}")

user1 = {"name":"Chetan" , "logged_in":True}
user2 = {"name":"Pratik" , "logged_in":False}
user3 = {"name":"Dhruv"  , "logged_in":True}
user4 = {"name":"Om"     , "logged_in":False}

view_profile(user1)
view_profile(user2)
view_profile(user3)
view_profile(user4)
```

# Multiple Decorators :
