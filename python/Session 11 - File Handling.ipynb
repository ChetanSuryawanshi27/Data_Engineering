# Types of data used for I/O :

- Text - '12345' as a sequence of unicode charts 
- Binary - 12345 as a sequence of bytes of its binary equvialent

Hence there are 2 file types to deal with 

- Text file - all program files are text files

# How File I/O done in most programming languages 

- Open a file 
- Read/Write data
- Close the file

# Case 1 if file is not present

```python
file = open("data.txt","w") # "w" = Writing Mode
file.write("Hello Chetan, How are you?")
file.close()
file.write("This is my first file")
```

# Write Multiple Line

```python
file = open("data.txt","w")         # "w" = writing mode / Overwrites Entire File
file.write("This is line no 1\n")   # \n  = New Line Charachter
file.write("This is line no 2\n")   
file.write("This is line no 3\n")
file.close()
```

# Write Using Writelines():

```python
lines = ["Hello Chetan\n", "How are you\n","You are the best coder\n"]

file = open("data.txt","w")
file.writelines(lines)
file.close()
```

# Append Mode : 

- "a" Mode - Append to end of file
- Append Mode ("a") adds data to end without deleting existing content

```python
file = open("Log.txt","w")
file.write("Event No 1\n")
file.close()
```

```python
file = open("Log.txt","a")  # "a" = Append Mode
file.write("Event No 2\n")
file.write("Event No 3\n")
file.close()
```

```python
file = open("Log.txt","a")
file.write("Event No 4\n")
file.close()
```

# Writing Many Files :

- create and write to multiple files

```python
files_data = {
    "file1.txt":"This is My Number 1 File\n",
    "file2.txt":"This is My Number 2 File\n",
    "file3.txt":"This is My Number 3 File\n"
}
```

```python
for filename , content in files_data.items():
    file = open(filename, "w")
    file.write(content)
    file.close()
```

# Reading Big File Line by Line :

- Use readline() for memory - efficient reading for large file

```python
file = open("data3.txt","w")
file.write("Python Program\n")
file.write("Java Program\n")
file.write("PHP Program\n")
file.write("C++ Program\n")
file.close()
```

```python
# Read one line at a time :

file = open("data3.txt","r")

while True:
    line = file.readline()
    if not line:
        break
    print(line.strip())

file.close()
```

```python
with open("Data3.txt","r") as file:
    for line in file:
        print(line.strip())
```

# Read Large file in fixed_size chunks to save memory :


```python
Big_L = ["Hello World" for i in range(1000)]
print(Big_L)
```

```python
with open("big.txt","w") as f:
    f.writelines(Big_L)
```

```python
with open("big.txt","r") as f:
    chunk_size = 100
    while len(f.read(chunk_size))>0:
        print(f.read(chunk_size))
        f.read(chunk_size)
```

# Problems with Text File :

# Create Gmail_Icon copy :

```python
with open("Gmail_Icon.png",'rb') as f:             # 'rb' = read binary
    with open("Gmail_Icon_copy.png",'wb') as wf:     # 'wb' = write binary
        wf.write(f.read())
```

# Serialization & Deserialization :

- Serialization : convert python object into json format
- Deserialization : convert json format into python object

# Serialization :

# Serialization using json module - dump()

### json.dump() writes json object to file in JSON format

#### JSON - Java Script Object Notation :

```python
import json
```

```python
# Python Object :

student = {
    "name":"Chetan",
    "age":25,
    "marks":[75,85,95]
}
```

```python
with open("student.json","w") as file:
    json.dump(student,file)
```

# Deserialization using json module - Load()

- json.load() reads JSON file and convert to python object

```python
import json
```

```python
with open("student.json","r") as file:
    student = json.load(file)
    print(student)
    print(type(student))
```

# Serialization & Deserialization of Tuple :

- JSON converts tuples to list

```python
import json
```

```python
coordinates = (10,20,30)

with open("coords.json","w") as file:
    json.dump(coordinates,file)
```

```python
# Deserialization of Tuple :

with open("coords.json","r") as file:
    loaded = json.load(file)
    print(loaded)
    print(type(loaded))
```

# Serialization and Deserialization of custom object :

- JSON doesn't serialize custom object directly

```python
# Serialization  custom object

class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
```

```python
student = Student("Chetan",23)
```

```python

with open("student.json","w") as file:
    json.dump(student,file)
```

```python
def show_object(student):
    if isinstance(student,Student):
        return "My Name is {} and I Am {} Years old".format(student.name,student.age)
```

```python
# Serialization of custom object  :

with open("student.json","w") as file:
    json.dump(student,file,default = show_object)
```

```python
# Deserialization of custom object  :

import json
```

```python
with open("student.json","r") as file:
    d = json.load(file)
    print(d)
    print(type(d))
```

# Pickling : 

- Pickle Serialize Python Objects Preserving Exact Type

```python
import pickle
```

```python
data = {
    "list":[1,2,3],
    "tuple":(4,5,6),
    "dict":{"a":1, "b":2}
}
```

# Serialization with Pickle :

```python
with open("data.pkl","wb") as file:
    pickle.dump(data,file)
```

# Derialization with Pickle :

```python
with open("data.pkl","rb") as file:
    loaded = pickle.load(file)
    print(loaded)
    print(type(loaded))
```
