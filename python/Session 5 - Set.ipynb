# 4) Deleting items in set 

- remove(item) - remove item   (error if item not found)
- discard(item) - remove item  ( no error if item not found)
- pop() - remove and return random item
- clear() - remove all item
- del - Delete entire set

```python
# remove method

fruits = {"Apple", "Banana","Mango","Orange"}

fruits.remove("Banana")
print(fruits)
```

```python
fruits.remove("Grapes")
```

```python
fruits = {"Apple", "Banana","Mango","Orange"}

fruits.discard("Apple")
print(fruits)
```

```python
fruits.discard("Banana")
print(fruits)
```

```python
fruits.discard("Grapes")
print(fruits)
```

```python
# pop() method

numbers = {1,2,3,4,5}

remove = numbers.pop()
print(remove)
print(numbers)
```

```python
# clear method

numbers = {100,200,300,400,500}

numbers.clear()
print(numbers)
```

```python
# del keyword

my_set = {1,2,3,4,5}

del my_set
```

# 5) set Operation 

- Union ( | ) , union() 
- intersection ( & ) , intersection()
- differnce ( - ) , differnce()
- symmetric differnce ( ^ ) , symmetric_differnce()
  

```python
set1 = {1,2,3,4}
set2 = {3,4,5,6}

result = set1 | set2
print(result)
```

```python
result = set1.union(set2)
print(result)
```

```python
# intersection (Common item)

set1 = {1,2,3,4}
set2 = {3,4,5,6}

result = set1 & set2
print(result)
```

```python
result = set1.intersection(set2)
print(result)
```

```python
# differnce (items in first set only)

set1 = {1,2,3,4}
set2 = {3,4,5,6}

result = set1 - set2
print(result)
```

```python
result = set1.difference(set2)
print(result)
```

```python
# symmetric differnce  (not in both)

set1 = {1,2,3,4}
set2 = {3,4,5,6}

result = set1 ^ set2 
print(result)
```

```python
result = set1.symmetric_difference(set2)
print(result)
```

- sum() - sum calculation
- sorted() - return sorted list

# 8) set coperhension

- concise way to create sets using a single line of code

- syntax {expression for item in iterable if condition}


```python
squre = set ()

for x in range(1,6):
    squre.add(x**2)

print(squre)
```

```python
squres = {x**2 for x in range(1,6)}
print(squres)
```

```python
evens = {x for x in range(1,51) if x % 2 == 0}

print(evens)
```

```python
div_by_3 = {x for x in range(1,21) if x % 3 == 0 }
print(div_by_3)
```
